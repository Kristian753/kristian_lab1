```python
import matplotlib.pyplot as plt

def analyze_dna_file(filename):
    # öppna fil
    with open(filename, 'r') as f:
        all_lines = f.readlines()  # läser alla rader
    
    sequences = {}  # dictionary för dna
    current_id = ""  # nuvarande id

    for line in all_lines:
        line = line.strip()  # rensa rad
        
        if line.startswith('>'):  # kollar id-rad
            current_id = line[1:]  # sparar id
            sequences[current_id] = ""  # skapar tom sträng
        else:  # om dna-rad
            if current_id:  # kolla att id finns
                sequences[current_id] += line.upper()  # gör bokstäver till versaler

    for name, sequence in sequences.items():
        count = {'A': 0, 'T': 0, 'C': 0, 'G': 0}  # starta räkna
        
        for letter in sequence:  # går igenom bokstäver
            if letter in count:  # kollar dna-bokstäver
                count[letter] += 1  # ökar räknare
        
        print(f"{name}: {count}")  # printar resultat
        
        colors = ['red', 'blue', 'green', 'yellow']  # färger på staplarna
        plt.bar(count.keys(), count.values(), color=colors)  # rita diagramen
        plt.title(name)  # lägger till titel
        plt.show()  # visa diagramen

    print(f"Analyserat klart! Sequences: {len(sequences)}")  # sista meddelande

# Använd funktionen med båda filerna:
analyze_dna_file('dna_raw.txt')
analyze_dna_file('dna_raw_complicated.txt')
```
