```python
import matplotlib.pyplot as plt

# Öppnar fil för läsning
f = open('dna_raw.txt', 'r')
alla_rader = f.readlines()  # Läser alla rader
f.close()  # Stänger filen

dna_sekvenser = {}  # Dictionary för DNA
id_nu = ""  # Nuvarande ID

for rad in alla_rader:
    rad = rad.strip()  # Rensar rad

 if rad.startswith('>'):  # Kollar ID-rad
        id_nu = rad[1:]  # Sparar ID
        dna_sekvenser[id_nu] = ""  # Skapar tom sträng
    else:  # Om DNA-rad
        if id_nu:  # Kollar att ID finns
            dna_sekvenser[id_nu] += rad.upper()  # Lägger till bokstäver


```
